```python

import pandas as pd
import numpy as np

np.random.seed(42)

campaigns = pd.DataFrame({
    "campaign_id": range(1, 3001),
    "marketing_spend": np.random.randint(500, 10000, 3000),
    "audience_size": np.random.randint(1000, 100000, 3000),
    "engagement_rate": np.random.uniform(0.01, 0.15, 3000).round(4)
})

campaigns["campaign_revenue"] = (
    campaigns["marketing_spend"] * 2.5
    + campaigns["audience_size"] * 0.3
    + campaigns["engagement_rate"] * 50000
    + np.random.normal(0, 3000, 3000)
).round(2)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, root_mean_squared_error, mean_absolute_error

X = campaigns[["marketing_spend","audience_size","engagement_rate"]]
y = campaigns["campaign_revenue"]

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.30, random_state =25)

model = LinearRegression ()
model.fit(X_train,y_train)
predictions = model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test,predictions)
rmse = root_mean_squared_error(y_test,predictions)
r2 = r2_score(y_test,predictions)

print(mae)
print(rmse)
print(r2)

print(model.coef_)
print(model.intercept_)
```

```python
new_campaign = pd.DataFrame({"marketing_spend":[5000],"audience_size":[45000],"engagement_rate":[0.08]})
new_campaign_revenue = model.predict(new_campaign)
print(new_campaign_revenue)
```
