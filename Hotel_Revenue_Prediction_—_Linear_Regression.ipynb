```python

import pandas as pd
import numpy as np

np.random.seed(42)

hotels = pd.DataFrame({
    "hotel_id": range(1, 2001),
    "monthly_bookings": np.random.randint(100, 1000, 2000),
    "average_nightly_rate": np.random.randint(70, 300, 2000),
    "review_score": np.random.uniform(3, 5, 2000).round(1)
})

hotels["monthly_revenue"] = (
    hotels["monthly_bookings"] * 120
    + hotels["average_nightly_rate"] * 200
    + hotels["review_score"] * 5000
    + np.random.normal(0, 8000, 2000)
).round(2)

hotels.head()
```

```python
hotels.head()
```

```python
hotels.shape
hotels.dtypes
hotels["monthly_revenue"].mean()
hotels.isna().sum()
hotels.describe()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, root_mean_squared_error

X = hotels[["monthly_bookings","review_score","average_nightly_rate"]]
y = hotels["monthly_revenue"]

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2, random_state = 40)
```

```python
model = LinearRegression()
model.fit (X_train,y_train)
predictions = model.predict (X_test)

mae = mean_absolute_error (y_test,predictions)
rmse = root_mean_squared_error(y_test,predictions)
r2 = r2_score (y_test,predictions)
```

```python
print (mae)
print (rmse)
print (r2)

print(model.coef_)
print(model.intercept_)
```

```python
new_hotel = pd.DataFrame ({"monthly_bookings":[150],"review_score":[4],"average_nightly_rate":[140]})

new_hotel_revenue = model.predict(new_hotel)

print(new_hotel_revenue)
```
